```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
category = pd.read_csv("../data/raw/05_category_inflows.csv")

# Convert month column
category["month"] = pd.to_datetime(category["month"])

# Format month for display
category["month"] = category["month"].dt.strftime("%Y-%m")

# Create pivot table
heatmap_data = category.pivot(
    index="category",
    columns="month",
    values="net_inflow_crore"
)

# Plot heatmap
plt.figure(figsize=(14,8))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".0f",
    cmap="YlGnBu",
    linewidths=0.5
)

plt.title("Category-wise Net Inflow Heatmap", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Mutual Fund Category")

plt.tight_layout()

# Save chart
plt.savefig(
    "../charts/category_inflow_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# ==================================
# Load NAV Files
# ==================================

icici = pd.read_csv("../data/raw/ICICI_Bluechip_NAV.csv")
kotak = pd.read_csv("../data/raw/Kotak_Bluechip_NAV.csv")
nippon = pd.read_csv("../data/raw/Nippon_LargeCap_NAV.csv")
sbi = pd.read_csv("../data/raw/SBI_Bluechip_NAV.csv")

# ==================================
# Rename NAV Columns
# ==================================

icici = icici.rename(columns={"nav": "ICICI"})
kotak = kotak.rename(columns={"nav": "Kotak"})
nippon = nippon.rename(columns={"nav": "Nippon"})
sbi = sbi.rename(columns={"nav": "SBI"})

# ==================================
# Convert Date Columns
# ==================================

for df in [icici, kotak, nippon, sbi]:
    df["date"] = pd.to_datetime(
        df["date"],
        dayfirst=True
    )

# ==================================
# Merge NAV Data
# ==================================

merged = icici[["date", "ICICI"]]

merged = merged.merge(
    kotak[["date", "Kotak"]],
    on="date",
    how="inner"
)

merged = merged.merge(
    nippon[["date", "Nippon"]],
    on="date",
    how="inner"
)

merged = merged.merge(
    sbi[["date", "SBI"]],
    on="date",
    how="inner"
)

# ==================================
# Sort by Date
# ==================================

merged = merged.sort_values("date")

# ==================================
# Calculate Daily Returns
# ==================================

returns = merged.set_index("date").pct_change()

returns = returns.dropna()

# ==================================
# Correlation Matrix
# ==================================

corr_matrix = returns.corr()

print("Correlation Matrix:")
print(corr_matrix)

# ==================================
# Plot Heatmap
# ==================================

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="YlGnBu",
    linewidths=0.5
)

plt.title(
    "Correlation Matrix of Daily NAV Returns (4 Mutual Funds)",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

# ==================================
# Save PNG
# ==================================

plt.savefig(
    "../charts/correlation_matrix_4funds.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

 8A. Correlation Matrix of NAV Returns

The four selected large-cap mutual funds exhibit strong positive correlations in daily NAV returns, indicating that they are influenced by similar market conditions and benchmark movements. High correlation values suggest limited diversification benefits when investing exclusively in funds from the same category.

Supporting Chart: `charts/correlation_matrix_4funds.png`

```python
import pandas as pd

nav = pd.read_csv("../data/raw/02_nav_history.csv")

print(nav.columns.tolist())
print(nav.head())
print("Unique AMFI Codes:", nav["amfi_code"].nunique())
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ==================================
# Load Datasets
# ==================================

nav = pd.read_csv("../data/raw/02_nav_history.csv")
fund_master = pd.read_csv("../data/raw/01_fund_master.csv")

# ==================================
# Convert Date
# ==================================

nav["date"] = pd.to_datetime(nav["date"])

# ==================================
# Select 10 Funds
# ==================================

selected_funds = nav["amfi_code"].unique()[:10]

nav_10 = nav[nav["amfi_code"].isin(selected_funds)]

# ==================================
# Get Scheme Names
# ==================================

scheme_map = (
    fund_master[["amfi_code", "scheme_name"]]
    .drop_duplicates()
)

# Merge Scheme Names
nav_10 = nav_10.merge(
    scheme_map,
    on="amfi_code",
    how="left"
)

# ==================================
# Create Pivot Table
# ==================================

pivot_nav = nav_10.pivot(
    index="date",
    columns="scheme_name",
    values="nav"
)

# ==================================
# Calculate Daily Returns
# ==================================

returns = pivot_nav.pct_change().dropna()

# ==================================
# Correlation Matrix
# ==================================

corr_matrix = returns.corr()

# ==================================
# Plot Heatmap
# ==================================

plt.figure(figsize=(12,10))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="YlGnBu",
    linewidths=0.5
)

plt.title(
    "Correlation Matrix of NAV Returns (10 Selected Mutual Fund Schemes)",
    fontsize=14,
    fontweight="bold"
)

plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

plt.tight_layout()

# ==================================
# Save Chart
# ==================================

plt.savefig(
    "../charts/correlation_matrix_10funds.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ==================================
# Load Dataset
# ==================================

holdings = pd.read_csv("../data/raw/09_portfolio_holdings.csv")

# ==================================
# Aggregate Sector Weights
# ==================================

sector_weights = (
    holdings.groupby("sector")["weight_pct"]
    .sum()
    .sort_values(ascending=False)
)

# Top 8 sectors for better visualization
top_sectors = sector_weights.head(8)

# Combine remaining sectors into 'Others'
others = sector_weights.iloc[8:].sum()

if others > 0:
    top_sectors["Others"] = others

# ==================================
# Create Donut Chart
# ==================================

plt.figure(figsize=(10, 8))

wedges, texts, autotexts = plt.pie(
    top_sectors.values,
    labels=top_sectors.index,
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.8
)

# Donut hole
centre_circle = plt.Circle(
    (0, 0),
    0.65,
    fc="white"
)

plt.gca().add_artist(centre_circle)

# ==================================
# Formatting
# ==================================

plt.title(
    "Sector Allocation Across Mutual Fund Portfolios",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

# ==================================
# Save Chart
# ==================================

plt.savefig(
    "../charts/sector_allocation.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
