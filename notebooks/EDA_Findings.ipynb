10 Key EDA Findings – Mutual Fund Analysis

 1. Strong Growth in NAV Trends

    NAV values show a consistent upward movement over time, indicating overall positive fund performance and market appreciation.

 2. Increasing AUM Reflects Investor Confidence

    Assets Under Management (AUM) exhibit steady growth, suggesting rising investor trust and continuous capital inflow into mutual fund schemes.

 3. SIP Inflows Show Strong Adoption

    SIP inflows demonstrate a sustained increasing trend, highlighting the growing preference for disciplined, long-term investment strategies.

 4. Rising Investor Participation (Folio Growth)

    The continuous increase in folio count indicates expanding investor base and greater penetration of mutual fund products.

 5. Active SIP Accounts Are Growing

    The growth in active SIP accounts shows increasing engagement in systematic investment plans among retail investors.

 6. Middle-Aged Investors Dominate Market

    Most investments come from middle-aged individuals, indicating that stable income and financial maturity drive investment behavior.

 7. Income Level Strongly Influences Investment

    Higher-income groups contribute significantly more to mutual fund investments, showing a clear correlation between income and investment capacity.

 8. Tier-1 Cities Lead Investments

    T30 cities dominate mutual fund investments, while B30 cities show gradual but steady growth, indicating increasing financial awareness in smaller regions.

 9. Sector Allocation Shows Diversification

    Investments are distributed across multiple sectors, with certain sectors receiving higher allocation based on risk-return preferences.

 10. Correlation Patterns Indicate Diversification Opportunity

     Correlation analysis shows both strongly correlated funds and independent movements, highlighting opportunities for portfolio diversification.

Summary:

Overall, the mutual fund ecosystem demonstrates strong growth, increasing SIP adoption, and expanding investor participation across demographics and regions.
